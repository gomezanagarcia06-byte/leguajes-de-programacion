### Tu tarea

Con `mi_portafolio_dict` y `mis_precios_dict`:

1. Construye un DataFrame con las columnas `ticker`, `cantidad`, `precio_compra`, `precio_actual`.
2. Agrega las columnas `valor_actual`, `ganancia` y `rendimiento`.
3. Filtra los activos en pérdida.
4. Ordena el DataFrame de mayor a menor rendimiento.

```python
import pandas as pd

# 1. Definición de diccionarios/datos basados en la imagen
mi_portafolio_dict = {
    "ticker": ["AMZN", "NVDA", "MSFT", "V"],
    "cantidad": [17, 8, 20, 10],
    "precio_compra": [143, 209, 212, 403]
}

mis_precios_dict = {
    "precio_actual": [144, 256, 181, 472]
}

df = pd.DataFrame(mi_portafolio_dict)
df["precio_actual"] = mis_precios_dict["precio_actual"]

df["valor_compra"] = df["cantidad"] * df["precio_compra"]
df["valor_actual"] = df["cantidad"] * df["precio_actual"]
df["ganancia"] = df["valor_actual"] - df["valor_compra"]
df["rendimiento"] = (df["ganancia"] / df["valor_compra"]) * 100

# 4. Filtrar los activos en pérdida (rendimiento < 0)
en_perdida = df[df["rendimiento"] < 0]

# 5. Ordenar de mayor a menor rendimiento
df_ordenado = df.sort_values(by="rendimiento", ascending=False)

# Imprimir resultados
print("--- Activos en pérdida ---")
print(en_perdida[["ticker", "cantidad", "ganancia", "rendimiento"]])

print("\n--- Portafolio ordenado por rendimiento (mayor a menor) ---")
print(df_ordenado)
```
