```python
AMZN = np.array([0.012, 0.008, -0.004, 0.015, 0.006, -0.002, 0.010, 0.003])

print("Número de días:", len(AMZN))
print("Primer día:", AMZN[0])
print("Último día:", AMZN[-1])
```

```python
promedio_aapl = AMZN.mean()
peor_dia = AMZN.min()

print("Promedio:", promedio_aapl)
print("Peor día:", peor_dia)
```

```python
riesgo_aapl = AMZN.std()
print("Riesgo (std):", riesgo_aapl)
```

```python
AMZN = np.array([0.012, 0.008, -0.004, 0.015, 0.006, -0.002, 0.010, 0.003])
NVDA = np.array([0.004, 0.003, 0.005, 0.002, 0.006, 0.004, 0.003, 0.005])
MSFT = np.array([-0.010, -0.015, 0.002, -0.008, -0.012, 0.005, -0.020, -0.006])
V = np.array([0.030, -0.025, 0.040, 0.015, -0.020, 0.035, -0.010, 0.020])
```

```python
activos = {"AMZN": AMZN, "NVDA": NVDA, "MSFT": MSFT, "V": V}

for nombre, serie in activos.items():
    promedio = serie.mean()
    riesgo = serie.std()
    print(f"{nombre}: rendimiento {promedio:.4f}  |  riesgo {riesgo:.4f}")
```

```python
pesos = np.array([0.35, 0.40, 0.10, 0.15])  # AMNZ, NVDA, MSFT, V — suman 1.0

rendimiento_portafolio = AMZN*pesos[0] + NVDA*pesos[1] + MSFT*pesos[2] + V*pesos[3]
print(rendimiento_portafolio)
```

```python
rendimiento_promedio_portafolio = rendimiento_portafolio.mean()
riesgo_portafolio = rendimiento_portafolio.std()

print(f"Rendimiento promedio del portafolio: {rendimiento_promedio_portafolio:.4f}")
print(f"Riesgo del portafolio: {riesgo_portafolio:.4f}")
```

```python
riesgos_individuales = np.array([AMZN.std(), NVDA.std(), MSFT.std(), V.std()])
riesgo_promedio_simple = (riesgos_individuales * pesos).sum()

print(f"Riesgo si solo promediaras los riesgos individuales: {riesgo_promedio_simple:.4f}")
print(f"Riesgo real del portafolio combinado: {riesgo_portafolio:.4f}")
```

```python
mis_pesos = np.array([0.25, 0.25, 0.25, 0.25])  # cambia estos números — deben sumar 1.0

mi_portafolio = AMZN*mis_pesos[0] + NVDA*mis_pesos[1] + MSFT*mis_pesos[2] + V*mis_pesos[3]
print(f"Rendimiento: {mi_portafolio.mean():.4f}  |  Riesgo: {mi_portafolio.std():.4f}")
```
